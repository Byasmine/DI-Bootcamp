```python
import pandas as pd

df = pd.read_csv('/content/Data Science Job Salary dataset/datascience_salaries.csv')
```

```python
print(df.head())
```

```python
print(df["salary"].head())

#min max salary value
print("Min Salary: ",df["salary"].min())
print("Max Salary: ",df["salary"].max())
```

```python
#Normalize the salary column
from sklearn.preprocessing import MinMaxScaler

min_max_scaler = MinMaxScaler()
df["salary_scale"] = min_max_scaler.fit_transform(df[["salary"]])

print(df["salary_scale"].head())
```

```python
#Dimensionality Reduction
from sklearn.decomposition import PCA

# Select numeric features (excluding target)
numeric_cols = df.select_dtypes(include=['float64','int64']).drop(columns=['salary']).columns

# Apply PCA to reduce to 2 components
pca = PCA(n_components=2)
df_pca = pca.fit_transform(df[numeric_cols])

print("Explained variance ratio:", pca.explained_variance_ratio_)
```

```python
# Average and median salary per experience level
salary_stats = df.groupby('experience_level')['salary'].agg(['mean','median'])
print(salary_stats)
```
