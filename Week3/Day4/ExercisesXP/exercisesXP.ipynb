### Exercise 2 : Creating a Line Plot for Temperature Variation

```python
import matplotlib.pyplot as plt

week = range(1, 8)
temps = [72, 74, 76, 80, 82, 78, 75]

#plot size
plt.figure(figsize=(10, 6))
plt.xlabel('Day')
plt.ylabel('Temperature (°F)')
plt.title('Temperature Variation Over a Week')
plt.plot(week, temps, marker='o', linestyle='-', color='b')
plt.grid(True , alpha=0.5)
```

```python
import matplotlib.pyplot as plt

sales = [5000, 5500, 6200, 7000, 7500]
month = ["january", "february", "march", "april", "may"]

plt.figure(figsize=(10, 6))
plt.xlabel('Month')
plt.ylabel('Sales Amount ($)')
plt.title('Sales Analysis')
plt.bar(month, sales, color='g' , alpha=0.7)
```

### Exercise 5: Comparing Anxiety Levels Across Different Genders

```python
import seaborn as sns
import matplotlib.pyplot as plt

df['Anxiety_binary'] = df['Do you have Anxiety?'].map({'Yes':1, 'No':0})

sns.barplot(
    data=df,
    x='Choose your gender',
    y='Anxiety_binary',
    color = 'pink'
)

plt.title("Anxiety Levels Across Different Genders")
plt.xlabel("Gender")
plt.ylabel("Anxiety Level")
```

### Exercise 6: Exploring the Relationship Between Age and Panic Attacks

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Convertir les réponses en numérique
df['Panic_binary'] = df['Do you have Panic attack?'].map({'Yes':1, 'No':0})

# Créer le scatter plot
plt.figure(figsize=(7,5))
sns.scatterplot(
    data=df,
    x='Age',
    y='Panic_binary',
    hue='Panic_binary',       # couleur différente pour Yes/No
    style='Panic_binary',     # style de point différent
    palette=['red','blue'],
    s=80                      # taille des points
)

# Personnaliser le graphique
plt.title("Relationship Between Age and Panic Attacks")
plt.xlabel("Age")
plt.ylabel("Panic Attacks (Yes=1, No=0)")
plt.yticks([0,1], ["No","Yes"])  # rendre l’axe Y plus lisible

plt.show()

```
