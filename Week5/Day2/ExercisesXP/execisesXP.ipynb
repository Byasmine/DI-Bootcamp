###  Exercise 1 : Matrix Operations

```python
import numpy as np

a = np.random.randint(0, 10, (3, 3))
print(a)
```

```python
#Determinant
det_A = np.linalg.det(a)
print("Determinant of A:", det_A)

#Inverse (only if determinant ≠ 0)
if det_A != 0:
    inv_A = np.linalg.inv(a)
    print("Inverse of A:\n", inv_A)
else:
    print("Matrix A has no inverse (determinant = 0).")
```

### Exercise 2 : Statistical Analysis

### Exercise 6 : Basic Hypothesis Testing

```python
import numpy as np

# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30).round(2)

# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30).round(2)

print(productivity_after)
```

```python
import numpy as np

# Generate the data (already given in your exercise)
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Step 1: Compute the averages
mean_before = np.mean(productivity_before)
mean_after = np.mean(productivity_after)

# Step 2: Print the results
print("Average before training:", mean_before)
print("Average after training:", mean_after)

# Step 3: Decision
if mean_after > mean_before:
    print("Alternative Hypothesis (H1): The training program improved productivity.")
else:
    print("Null Hypothesis (H0): The training program did NOT improve productivity.")
```

### Exercise 8 : Time Series Data Manipulation

```python
import numpy as np

# Full year 2023 daily time series
dates = np.arange('2023-01-01', '2024-01-01', dtype='datetime64[D]')
values = np.random.randint(10, 100, size=len(dates))

print("Length:", len(dates))
```

```python
# January to marche
jan_to_mar = values[
    (dates >= np.datetime64("2023-01-01")) &
    (dates <= np.datetime64("2023-03-31"))
]

print("January to March:\n",len(jan_to_mar) )
```

```python
#April to June
apr_to_jun = values[
    (dates >= np.datetime64("2023-04-01")) &
    (dates <= np.datetime64("2023-06-30"))
]
print("April to June:\n", len(apr_to_jun))
```

```python
#July to September
jul_to_sep = values[
    (dates >= np.datetime64("2023-07-01")) &
    (dates <= np.datetime64("2023-09-30"))
]
print("July to September:\n", len(jul_to_sep))
```

```python
#October to December
oct_to_dec = values[
    (dates >= np.datetime64("2023-10-01")) &
    (dates <= np.datetime64("2023-12-31"))
]
print("October to December:\n", len(oct_to_dec))
```

### Exercise 9 : Data Conversion

```python
import numpy as np
import pandas as pd

arry = np.random.randint(1, 101, (5, 4))

# Numpy > Pandas df
df = pd.DataFrame(arry, columns=['col1', 'col2', 'col3', 'col4'])

print("NumPy array:\n", arry)
print("\nDataFrame:\n", df)
```

```python
# Convert back to NumPy array
arr_back = df.to_numpy()
print("NumPy array:\n", arr_back)
```

### Exercise 10 : Basic Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 20 random numbers between 0 and 100
data = np.random.randint(0, 100, size=20)

print("Data:", data)

# bar chart
plt.bar(range(len(data)), data, color='orange')
plt.title("Bar Chart of Random Numbers")
plt.show()
```

```python
# Plot the data as a line graph
plt.plot(data, marker='o', linestyle='-', color='r', label='Random Data')

# Add labels and title
plt.xlabel("Index")
plt.ylabel("Value")
plt.title("Line Graph of Random Numbers")
plt.legend()

# Show the plot
plt.show()
```
