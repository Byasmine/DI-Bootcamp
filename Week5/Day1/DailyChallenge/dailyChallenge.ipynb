## Daily Challenge: NumPy, Pandas, and Matplotlib Integration

### 1. Data Preparation:

```python

import numpy as np
import pandas as pd

# Générer un tableau 10x12 (10 villes, 12 mois)
data = np.random.uniform(low=-5, high=35, size=(10, 12)).round(2)

# Noms des villes
cities = [f"City_{i+1}" for i in range(10)]

# Noms des mois
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

#Construire le DataFrame
df = pd.DataFrame(data, index = cities, columns = months)

print(df.head())
```

### 2. Data Analysis:

```python
#anual average temperature for each city
anual_avg_temp = df.mean(axis=1)
print(anual_avg_temp)

print("Highest Average Temperature: ", anual_avg_temp.idxmax( axis =0))

print("Lowest Average Temperature:", anual_avg_temp.idxmin(axis =0))
```

### 3. Data Visualization :

```python
import seaborn as sns
import matplotlib.pyplot as plt

#  températures de City_1
df.loc["City_1"].plot(kind="line", marker="o")
plt.title("Monthly Temperatures - City_1")
plt.xlabel("Months")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
#Comparer plusieurs villes
df.T.plot(figsize=(10,6))   # transpose pour avoir les mois en x-axis
plt.title("Temperature Trends Across Cities")
plt.xlabel("Months")
plt.ylabel("Temperature (°C)")
plt.legend(title="Cities", bbox_to_anchor=(1.05, 1))
plt.show()
```

```python
#Visualiser les moyennes annuelles
annual_avg = df.mean(axis=1)

plt.figure(figsize=(8,5))
sns.barplot(x=annual_avg.index, y=annual_avg.values, palette="viridis")
plt.title("Annual Average Temperature per City")
plt.ylabel("Temperature (°C)")
plt.xticks(rotation=45)
plt.show()
```

```python
#Heatmap

plt.figure(figsize=(10,6))
sns.heatmap(df, annot=True, cmap="coolwarm")
plt.title("Heatmap of Monthly Temperatures per City")
plt.show()
```
