```python
import pandas as pd

df= pd.read_csv("/content/drive/MyDrive/BootGenAI/wld_happiness_rep_2019.csv")
df.head()
```

```python
df.info()
```

```python
# Handle missing values
df = df.dropna()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plt.scatter(df["Social support"], df["Score"], color="teal", alpha=0.7)
plt.title("Social Support vs Happiness Score (2019)")
plt.xlabel("Social Support")
plt.ylabel("Happiness Score")
plt.grid(True, alpha=0.5)
plt.show()
```

```python
# Group by region (30)
region_data = df.groupby("Country or region")[["GDP per capita","Healthy life expectancy"]].mean().head(30)

# Plot
fig, ax1 = plt.subplots(figsize=(15,6))

# Bar plot for GDP
region_data["GDP per capita"].plot(kind="bar", ax=ax1, color="salmon", alpha=0.7)
ax1.set_ylabel("GDP per Capita", color="salmon")
ax1.set_title("GDP vs Healthy Life Expectancy by Region (2019)")

# Line plot for Healthy Life Expectancy
ax2 = ax1.twinx()
region_data["Healthy life expectancy"].plot(kind="line", ax=ax2, color="blue", marker="o")
ax2.set_ylabel("Healthy Life Expectancy", color="blue")


plt.grid(True, alpha=0.3)
plt.show()
```
