### Exercise 2 : Creating a Line Plot for Temperature Variation

```python
import matplotlib.pyplot as plt

week = range(1, 8)
temps = [72, 74, 76, 80, 82, 78, 75]

#plot size
plt.figure(figsize=(10, 6))
plt.xlabel('Day')
plt.ylabel('Temperature (°F)')
plt.title('Temperature Variation Over a Week')
plt.plot(week, temps, marker='*', linestyle='-', color='pink', markersize=12)
plt.grid(True , alpha=0.5)
```

### Exercise 3 : Visualizing Monthly Sales with a Bar Chart

```python
import matplotlib.pyplot as plt

months = range(1, 13)
salarys=[5000, 5500, 6200, 7000, 7500,5000,5500,6000,6700,5020,8000,7002]

# Liste des mois
month_names = ["January", "February", "March", "April", "May", "June",
               "July", "August", "September", "October", "November", "December"]

plt.figure(figsize=(10, 6))
plt.xlabel('Month')
plt.ylabel('Sales Amount ($)')
plt.title('Monthly Sales')
plt.bar(months, salarys, color='#B62651')

# Remplacer 1,2,3... par les noms des mois
plt.xticks(months, month_names, rotation=45)  # rotation pour lisibilité

plt.grid(True, alpha=0.5)
```

### Exercise 4 : Data Visualisation

```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/BootGenAI/sales_data.csv')

#transform object to timestamp
df['date'] = pd.to_datetime(df['date'])
print(df.info())
```

```python
#Total quantity of products sold
total_quantity = df["quantity"].sum()
print("Total quantity sold:", total_quantity)
```

```python
category_map = {
    "Clohting": "Clothing",
    "Shoeses": "Shoes",
    "Bgas": "Bags"
}

df["category"] = df["category"].replace(category_map)

print(df["category"].value_counts())
```

```python
#Highest revenue category
revenu_categories = df.groupby("category")["revenue"].sum()
highest_category = revenu_categories.idxmax()
highest_revenue = revenu_categories.max()

print("Highest revenue category:", highest_category)
print("Revenue generated:", highest_revenue)
```

```python
#Average revenue per sale
average_revenue = df["revenue"].mean()
print("Average revenue per sale:", average_revenue)
```

```python
#extract the quarter from the date (year)
df["quarter"] = df["date"].dt.quarter

#goup by quarter and sum revenue
quarter_revenue = df.groupby("quarter")["revenue"].sum()
print(quarter_revenue)

#plot the bar
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(quarter_revenue.index, quarter_revenue.values, color="#B62651")

# Customize labels
plt.xticks([1, 2, 3, 4], ["Q1", "Q2", "Q3", "Q4"])
plt.ylim(150000, 200000)  # limite de l'axe Yplt.title("Total Revenue per Quarter")
plt.xlabel("Quarter")
plt.ylabel("Revenue ($)")
plt.grid(True, alpha=0.5)
plt.show()
```

### Exercise 5 : Data Visualisation Using MatPlotLib

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(-10, 11)   # tableau NumPy
y = x**2


plt.figure(figsize=(10, 6))
plt.xlabel('X')
plt.ylabel(' Y')
plt.title('y = x^2')
plt.plot(x, y, linestyle='-', color='r', markersize=12)
plt.grid(True, alpha=0.5)
plt.show()
```

```python
products = ["A", "B", "C", "D"]
sales_val = [15,30,45,20]

plt.figure(figsize=(10, 6))
plt.title("Weekly Sales of Products")
plt.bar(x=products, height=sales_val, color='#B62651')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.grid(True, alpha=0.5)
plt.show()
```

```python
fruits = ["Apples", "Bananas","Cherries", "Dates"]
percentages = [40, 30, 20, 10]

plt.figure(figsize=(10, 6))
plt.title("Fruit Distribution")
plt.pie(x=percentages, labels=fruits , autopct="%1.1f%%",  shadow=True, explode=(0.1,0,0,0))
plt.show()
```
